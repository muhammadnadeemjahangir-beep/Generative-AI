```python
import os
from google.colab import userdata

# Load token and remove leading/trailing spaces and newlines
hf_token = userdata.get("HF_TOKEN").strip()

if not hf_token.startswith("hf_"):
    raise ValueError(
        "Invalid Hugging Face token. It must begin with 'hf_'."
    )

os.environ["HF_TOKEN"] = hf_token

print("Token loaded successfully.")
```

```python
from langchain_huggingface import ChatHuggingFace, HuggingFacePipeline

llm = HuggingFacePipeline.from_model_id(
    model_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task="text-generation",
    device=-1,  # Use CPU
    pipeline_kwargs={
        "max_new_tokens": 100,
        "do_sample": False,
        "return_full_text": False,
    },
)

model = ChatHuggingFace(llm=llm)

result = model.invoke("What is the capital of India?")

print(result.content)
```

```python
from groq import Groq
import gradio as gr
import os
```

```python
import getpass
from groq import Groq

api_key = getpass.getpass("Enter your Groq API key: ")

client = Groq(api_key=api_key)
```

```python
def summarize_text(text):

    if not text.strip():
        return "Please enter some text."

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "system",
                "content": "You are an expert text summarizer."
            },
            {
                "role": "user",
                "content": f"Summarize the following text in 5 bullet points:\n\n{text}"
            }
        ],
        temperature=0.3,
        max_tokens=300
    )

    return response.choices[0].message.content
```

```python
demo = gr.Interface(
    fn=summarize_text,
    inputs=gr.Textbox(
        lines=10,
        placeholder="Paste a long paragraph here..."
    ),
    outputs=gr.Textbox(
        label="Summary"
    ),
    title="📄 AI Text Summarizer",
    description="Summarize long text using the Groq API and Llama 3.3."
)
```

```python
demo.launch()
```
